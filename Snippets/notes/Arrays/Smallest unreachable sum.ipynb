You have n coins with positive integer values. What is the smallest sum you cannot create using a subset of the coins?

Learning point: find most COMPACT summary of what's achievable so far


Solution:
Sort the coins and keep track of R, the largest value such that every sum from 1 to R is buildable (starting at 0). For each coin x, if x > R + 1 then R + 1 is impossible (all remaining coins are too big to fill that gap), so print R + 1; otherwise add x to R, and if you get through every coin, the answer is the total sum plus 1.


Extra:
Why there are no gaps to subsequent cumul even if next coin is not R+1:

The key is that the invariant says every value from 1 to R is already buildable, not just the cumulative sum itself. So with R = 50, the values 40 through 50 are already buildable by definition, using only the earlier coins.
