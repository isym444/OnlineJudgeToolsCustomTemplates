Impossible but can do equivalent by using fenwick tree

Preprocess by doing coordinate compression

Every time instead of ms.emplace(x), do fw.add(compressed(x),1)

When you want to do indlb(ms, y), you do this:

fw.sum(0,compressed(y)) = sum of all elements smaller than y = index in sorted list of elements




N.B. to find compressed(z) you simply convert set of z values into vector and indlb(z) on that
