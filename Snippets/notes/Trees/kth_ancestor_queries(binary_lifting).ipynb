```cpp
int main()
{
  nyan;
  int n, q;
  cin >> n >> q;
  const int LOG = 18;                                 // 2^18 > 2e5
  vector<vector<int>> up(LOG, vector<int>(n + 1, 0)); // up[level][node], 0 = no ancestor

  for (int v = 2; v <= n; v++)
    cin >> up[0][v]; // level 0 = input

  for (int j = 1; j < LOG; j++)
    for (int v = 1; v <= n; v++)
      up[j][v] = up[j - 1][up[j - 1][v]]; // two hops of 2^(j-1)

  while (q--)
  {
    int x, k;
    cin >> x >> k;
    for (int j = 0; j < LOG && x; j++)
      if ((k >> j) & 1)
        x = up[j][x];
    cout << (x ? x : -1) << '\n';
  }
}
```
