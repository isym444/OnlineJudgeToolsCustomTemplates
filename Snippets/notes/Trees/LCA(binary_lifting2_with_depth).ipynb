BFS/DFS to get depth of each node

Use Binary lifting to lift the deeper node up to the depth of the shallower node (lift by depth[deeper] - depth[shallower]) 

If the nodes are now equal, that is the LCA.

If the nodes are not equal:
- Lift both together while distinct: 
    - Try all possible lifts starting from largest possible lift (height of tree), down to lift of 1
    - If lift causes a & b to be same node, ignore the lift (n.b. The 0 sentinel makes overshooting past the root safe, since both nodes become 0 and the jump is skipped)
    - If lift causes a & b to still be different nodes, update the values of a & b
- a and b are now the two children of the LCA, so the answer is up[0][a]. 

Total time is O((n + q) log n).
