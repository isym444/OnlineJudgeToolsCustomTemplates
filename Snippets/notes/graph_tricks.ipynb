Find whether exists and if so, print an odd cycle in a simple, connected, undirected graph:
- ODD cycle in a graph => graph is NOT bipartite i.e. cannot be coloured checkerboard
- So BFS until find node you have visited before AND not parent of cur node AND of same colour as cur node
- To print, just follow parents to root and as cycle odd numbered, you will reach a common ancestor
