method reference :: is the same as a lambda
```java
    public List<String> toUpperCase(List<String> words) {
        // TODO: write your implementation
        return words.stream().map(String::toUpperCase).toList();
    }

        public List<String> toUpperCase(List<String> words) {
        // TODO: write your implementation
        return words.stream().map(x->x.toUpperCase()).toList();
    }
```

here x.stream() is in lambda because at this point you are still dealing with streams and x is a List< Integer > so x.stream() converts it to Stream< Integer >

```java
    /**
     * Flatten a list of lists into one list, preserving order.
     */
    public List<Integer> flatten(List<List<Integer>> nested) {
        // TODO: write your implementation
        return nested.stream().flatMap(x -> x.stream()).toList();
    }
```

mapmulti is map except that for each input item you can output zero, one, or many items. In the code below, y is known as the sink or consumer

```java
    /**
     * Each element repeated 'times' times, in order: ("a","b"), 2 -> a,a,b,b. Try mapMulti.
     */
    public List<String> repeatEach(List<String> words, int times) {
        // TODO: write your implementation
        return words.stream().<String>mapMulti((x, y) -> {
            for (int i = 0; i < times; i++) {
                y.accept(x);
            }
        }).toList();
    }
```

Intstream.range(x,y) gives you an intstream with values in range x to y

```java
    /**
     * Prefix each element with its index: ["x","y"] -> ["0:x","1:y"]. (hint: IntStream.range + mapToObj)
     */
    public List<String> withIndex(List<String> words) {
        // TODO: write your implementation
        return IntStream.range(0, words.size()).mapToObj(x -> String.valueOf(x) + ":" + words.get(x)).toList();
    }
```

Sorting streams in descending order:
```java
    /**
     * Descending order.
     */
    public List<Integer> descending(List<Integer> numbers) {
        // TODO: write your implementation
        return numbers.stream().sorted(Comparator.reverseOrder()).toList();
    }
```

Lambdas for sorting streams:
```java
    /**
     * Sort by length; strings of equal length sorted alphabetically.
     */
    public List<String> byLengthThenAlphabetical(List<String> words) {
        // TODO: write your implementation
        return words.stream().sorted((a, b) -> {
            if (a.length() == (b.length())) {
                return a.compareTo(b);
            }
            return Integer.compare(a.length(), b.length());
        }).toList();
    }
```

Note sorting strings alphabetically uses a.compareTo(b)

To check whether all elements in a stream match a predicate:

```java
    /**
     * Is the list sorted ascending? (hint: compare neighbours using IntStream.range)
     */
    public boolean isSortedAscending(List<Integer> numbers) {
        // TODO: write your implementation
        return IntStream.range(0, numbers.size() - 1).allMatch(x -> {
            return numbers.get(x) <= numbers.get(x + 1);
        });
    }
```
